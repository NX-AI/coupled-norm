```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
import plotly.express as px
import pandas as pd

model_id = "Qwen/Qwen3-0.6B" # "Qwen/Qwen3-0.6B" # "gpt2", "meta-llama/Llama-3.2-1B-Instruct", "Qwen/Qwen3-1.7B" 

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, output_hidden_states=True, device_map="auto")
```

```python
def rms_mean(x):
  return (x.to(torch.float32).pow(2).mean(dim=-1, keepdim=True) + 1e-6).sqrt()

pre_rmsnorm_rms = {}

def get_pre_norm_hook(module_name):
    def pre_hook(module, args):
        hidden_state = args[0]
        pre_rmsnorm_rms[module_name] = rms_mean(hidden_state).cpu()
    return pre_hook

hook_handles = []
for name, module in model.named_modules():
    norm_keys = ["post_attention_layernorm", "input_layernorm", "model.norm", "ln_1", "ln_2", "ln_f"]
    if any([key in name for key in norm_keys]):
        handle = module.register_forward_pre_hook(get_pre_norm_hook(name))
        hook_handles.append(handle)

print(f"Attached pre-hooks to {len(hook_handles)} RMSNorm layers.")

inputs = tokenizer(prompt, return_tensors="pt", truncation=True).to(model.device)

with torch.no_grad():
    outputs = model(**inputs)

print("\nCaptured prenorm RMS:")
for name, tensor in pre_rmsnorm_rms.items():
    print(f"RMS: {tensor} | Module: {name}")

for handle in hook_handles:
    handle.remove()

layer_count = len(pre_rmsnorm_rms.values())
```

```python
layer_count = len(pre_rmsnorm_rms.values())
layer_names = [(f"A{i//2}" if i % 2 == 0 else f"M{i//2}") if i != layer_count - 1 else "OUT" for i in range(layer_count)]

df = pd.DataFrame({"model_id": clean_model_ids(model_id), "layer": layer_names, "rms": [t.mean().item() for k, t in pre_rmsnorm_rms.items()]})

fig = px.line(df, x='layer', y='rms', markers=True, labels={"rms": "Mean RMS", "layer": "Layer", "model_id": "Model"})
y_range = None if not "llama" in model_id else [-2, 0]
fig.update_yaxes(type="log", range=y_range)

fig.update_xaxes(tickmode='array', tickvals=df['layer'], ticktext=df['layer'], tickangle=90)

for i in range(1, len(df), 2):
    fig.add_vrect(
        x0=i - 1, 
        x1=i + 0,
        fillcolor="red", 
        opacity=0.1,
        layer="below",
        line_width=0,
    )

fig.update({**paper_style_dict, 'layout.width': 1200, 'layout.legend.y': 0, 'layout.margin.b': 100})
fig.write_image(f"norm-{clean_model_ids(model_id)}.pdf")
fig.show()
```
